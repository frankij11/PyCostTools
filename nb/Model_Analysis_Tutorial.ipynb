## 2. Creating a Sample Dataset

For this tutorial, we'll create a synthetic dataset representing project cost factors. This will allow us to demonstrate the modeling capabilities without requiring external data.

Our dataset will simulate a typical cost estimation scenario with the following features:
- `project_size`: Size of the project in square feet
- `material_cost`: Cost per unit of material
- `labor_hours`: Required labor hours
- `complexity`: Categorical variable indicating project complexity (Low, Medium, High)
- `region`: Geographic region (North, South, East, West)
- `year`: Year the project is planned

The target variable will be `total_cost`, which is calculated based on the features with some noise added to simulate real-world variability.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Generate sample data - simulating project cost factors
n_samples = 100
data = {
    'project_size': np.random.uniform(10, 1000, n_samples),  # Size in square feet
    'material_cost': np.random.uniform(5, 30, n_samples),    # Cost per unit
    'labor_hours': np.random.uniform(20, 200, n_samples),    # Hours
    'complexity': np.random.choice(['Low', 'Medium', 'High'], n_samples),
    'region': np.random.choice(['North', 'South', 'East', 'West'], n_samples),
    'year': np.random.randint(2015, 2024, n_samples)
}

# Create DataFrame
df = pd.DataFrame(data)

# Generate the target variable (total_cost) with some noise
# Here we use a simple model where:
# - total_cost depends on project_size * material_cost (materials)
# - plus labor_hours * hourly rate (labor)
# - multiplied by a complexity factor
# - with 10% random noise to simulate real-world variability
complexity_factor = {'Low': 1.0, 'Medium': 1.3, 'High': 1.8}
df['complexity_factor'] = df['complexity'].map(complexity_factor)
df['total_cost'] = (
    df['project_size'] * df['material_cost'] + 
    df['labor_hours'] * 50 * df['complexity_factor']
) * (1 + np.random.normal(0, 0.1, n_samples))  # Add 10% noise

# Display the first few rows of the dataset
df.head()
```

Before building models, let's explore our dataset to get a better understanding of the data.

## 3. Single Model Analysis with the `Model` Class

The `Model` class provides a unified interface for building and evaluating cost models. It handles:

1. Data preprocessing (handling missing values, scaling features)
2. Formula-based feature selection using patsy syntax (similar to R formulas)
3. Model fitting and evaluation
4. Prediction on new data
5. Model persistence (saving/loading models)
6. Reporting and visualization

Let's start by building a simple linear regression model to predict total project cost.

### Creating and Fitting a Model

The `Model` class accepts several important parameters:

- `df`: The DataFrame containing our data
- `formula`: A patsy-style formula string defining the relationship between features and target
- `model`: The scikit-learn model to use (default is RandomForestRegressor)
- `test_split`: Proportion of data to use for testing (default is 0.2 or 20%)
- `random_state`: For reproducibility
- Metadata parameters like `title`, `description`, and `analyst` for documentation

```python
# Create and fit a Model
model = Model(
    df=df,
    formula="total_cost ~ project_size + material_cost + labor_hours + complexity + region-1",
    model=LinearRegression(),
    test_split=0.2,
    random_state=42,
    title="Project Cost Analysis",
    description="Predicting total project cost based on various factors",
    analyst="Data Scientist"
)

# Fit the model
model.fit()
```

### Model Summary and Evaluation

After fitting the model, we can get a summary of its performance metrics. This includes:

- R-squared (RSQ): Proportion of variance explained
- Mean Squared Error (MSE): Average squared difference between predicted and actual values
- Absolute Error (AbsErr): Average absolute difference between predicted and actual values
- Coefficient of Variation (CV): Standard deviation relative to the mean
- Maximum Error: Largest prediction error

```python
# Print model summary
model_summary = model.summary()
model_summary
```

### Making Predictions

Now that we have a fitted model, we can use it to make predictions on new data or on our existing data.

```python
# Make predictions
df = df.assign(linear_predictions = model.predict(df))
df.plot(x='linear_predictions', y='total_cost', kind='scatter')
df.head()
```

### Model Coefficients

For linear models, we can examine the coefficients to understand the relationship between features and the target.

```python
pipe = model.model
data= model.df.head()
def show_transformation(pipe, data):
    results = {}

    for step in pipe.steps:
        if hasattr(step[1], 'steps'):
            show_transformation(step[1], data)


        if hasattr(step[1], 'transform'):
            print(step[0])
            data = step[1].transform(data)
            print(data)
        elif hasattr(step[1], 'predict'):
            print(step[0])
            y = step[1].predict(data)
            print(y)
show_transformation(pipe, data)
```

### Visualizing Model Performance

Let's visualize the actual vs. predicted values to assess our model's performance.

```python
# Predicted vs. Actual Plot
y_pred = model.predict(model.X_test)
y_test = model.y_test

plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min()[0], y_test.max()[0]], 
         [y_test.min()[0], y_test.max()[0]], 
         'r--')
plt.xlabel('Actual Cost')
plt.ylabel('Predicted Cost')
plt.title('Actual vs Predicted Cost')
plt.grid(True)
plt.show()
```

## 4. Multiple Models Comparison with the `Models` Class

The `Models` class allows us to build and compare multiple models simultaneously, with different:
- Formulas (feature combinations)
- Model types (algorithms)
- Data subsets (by grouping variables)

This is extremely useful for model selection and can save significant time compared to manual model building and comparison.

### Define Model Variations

Let's define different formulas to test, as well as different model types.

```python
# Create different formulas to test
formulas = [
    "total_cost ~ project_size + material_cost",  # Basic materials model
    "total_cost ~ project_size + material_cost + labor_hours",  # Materials + labor model
    "total_cost ~ project_size + material_cost + labor_hours + complexity"  # Full model without region
]

# Create different models to compare
model_types = [
    LinearRegression(),  # Simple linear regression
    RidgeCV(alphas=np.logspace(-3, 3, 10)),  # Ridge regression with cross-validation
    LassoCV(alphas=np.logspace(-3, 3, 10)),  # Lasso regression with cross-validation
    RandomForestRegressor(n_estimators=100, random_state=42)  # Random forest
]
```

### Create and Fit Multiple Models

Now we'll use the `Models` class to create and fit all combinations of formulas and model types.

```python
# Create a Models instance to compare different model combinations
models = Models(
    df=df,
    formulas=formulas,
    models=model_types,
    test_split=0.2,
    random_state=42,
    title="Project Cost Model Comparison",
    description="Comparing different model types and formulas",
    analyst="Data Scientist"
)

# Fit all models (with a timeout in case some models take too long)
models.fit(timeout_in_seconds=60)
```

### Compare Model Performance

After fitting all models, we can compare their performance to select the best one.

```python
# Get summary of all models
summary = models.summary()

# Display key metrics sorted by R-squared (descending)
comparison = summary[['Formula', 'Model', 'RSQ', 'MSE', 'AbsErr']].sort_values('RSQ', ascending=False)
comparison
```

### Visualize Model Comparison

Let's create a visual comparison of model performance.

```python
# Create a bar chart of R-squared values for each model
plt.figure(figsize=(14, 6))

# Create labels combining formula and model type
labels = [f"{row['Formula'].split('~')[1][:20]}...\n{str(row['Model'])[:25]}" 
          for _, row in comparison.iterrows()]

plt.bar(range(len(comparison)), comparison['RSQ'], color='steelblue')
plt.xticks(range(len(comparison)), labels, rotation=90)
plt.xlabel('Model')
plt.ylabel('R-squared')
plt.title('Model Performance Comparison (R-squared)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## 5. Model Persistence

## 8. Advanced Features

### 8.1 Implementing Custom Preprocessing

The `Model` class allows you to customize the preprocessing pipeline to handle specific data requirements.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

# Example of a custom preprocessor
numeric_features = ['project_size', 'material_cost', 'labor_hours']
categorical_features = ['complexity', 'region']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

custom_preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# Create a model with custom preprocessing
custom_model = Model(
    df=df,
    formula="total_cost ~ project_size + material_cost + labor_hours + complexity + region",
    model=RandomForestRegressor(n_estimators=100, random_state=42),
    preprocessor=custom_preprocessor,
    test_split=0.2,
    random_state=42,
    title="Custom Preprocessed Model"
)

# Fit the model with custom preprocessing
custom_model.fit()

# Get model performance
custom_model.summary()
```

### 8.2 Model Interpretation and Feature Importance

For tree-based models like Random Forest, we can extract feature importance to understand which factors most influence the cost.

```python
# Create and fit a Random Forest model
rf_model = Model(
    df=df,
    formula="total_cost ~ project_size + material_cost + labor_hours + complexity + region",
    model=RandomForestRegressor(n_estimators=100, random_state=42),
    test_split=0.2,
    random_state=42,
    title="Random Forest Model"
)

rf_model.fit()

# Extract feature importance
try:
    importances = rf_model.model['model'].feature_importances_
    feature_names = rf_model.feature_cols
    
    # Create a DataFrame to display importance
    importance_df = pd.DataFrame({
        'Feature': feature_names,
        'Importance': importances
    }).sort_values('Importance', ascending=False)
    
    # Visualize feature importance
    plt.figure(figsize=(10, 6))
    plt.barh(importance_df['Feature'], importance_df['Importance'], color='skyblue')
    plt.xlabel('Importance')
    plt.title('Feature Importance (Random Forest)')
    plt.grid(axis='x', linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
    
    importance_df
except Exception as e:
    print(f"Could not extract feature importance: {e}")
```

## 9. Tips and Best Practices

Here are some recommendations for getting the most out of the PyCostTools modeling classes:

1. **Start Simple**: Begin with a simple model using only a few key features before adding complexity.

2. **Use the Models Class for Model Selection**: The `Models` class is great for quickly comparing different modeling approaches and feature sets.

3. **Document Your Analysis**: Use the metadata parameters (`title`, `description`, `analyst`) to document your analysis for future reference.

4. **Handle Categorical Variables Carefully**: Categorical variables can be included directly in the formula, but for more control, consider custom preprocessing.

5. **Save Your Models**: Always save your trained models to avoid having to retrain them. This is especially important for models that take a long time to train.

6. **Use Visual Reports**: The report functionality provides valuable insights into model performance and can help communicate results to stakeholders.

7. **Check Model Performance on Test Data**: Always evaluate your model's performance on held-out test data to ensure it generalizes well.

8. **Consider Feature Engineering**: Although the `Model` class handles basic preprocessing, consider engineering new features that capture domain knowledge about cost drivers.
